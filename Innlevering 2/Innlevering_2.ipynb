## Rekursjon

En rekursjon er gitt ved

$u_{n+1}=u_n+f(u_n)$

Følgen $\{u_n\}_{n=0}^{\infty}$ er gitt ved en initialverdi $u_0$ og en rekursjon.

## Diskret logistisk ligning

Den diskrete logistiske ligningen er

$u_{n+1}=u_n+ru_n(K-u_n)$

der

$r=$ vekstrate

$K=$ bærekapasitet

$u_0=$ initialverdi



```python
# pakkene vi trenger
%matplotlib inline
import numpy as np # trengs senere: sinus-funksjonen er np.sin()
import matplotlib.pyplot as plt

# høyresida i rekursjonen
def f(x,r):
    return (1+r)*x

# beregner følgen
def beregn_folge(
r, # vekstfaktor
u0, # initialverdi
nmax # antall iterasjoner
):
    u = np.zeros(nmax+1) # vi begynner med en vektor som har nuller overalt
    u[0] = u0
    for i in range(0, nmax):
        u[i+1] = f(u[i],r)

    return u # returnerer en vektor som representerer følgen

# lager plot
def plott_folge(u):
    naturlige_tall = np.arange(0.0, len(u), 1)
    # fikser litt på pyplot
    newparams = {'figure.figsize': (8.0, 4.0), 'axes.grid': True,
    'lines.markersize': 8, 'lines.linewidth': 2,
    'font.size': 14}
    plt.rcParams.update(newparams)
    plt.plot(naturlige_tall, u, 'bo') # plotter følgen for gitt vektor u

u_1 = beregn_folge(-1/2,1,100)
u_2 = beregn_folge(0,1,100)
u_3 = beregn_folge(1/2,1,100)


plott_folge(u_1)
plott_folge(u_2)
plott_folge(u_3)

print(u_1)
print(u_2)
print(u_3)
```
